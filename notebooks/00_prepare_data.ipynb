```python
!pip install -q beir
from beir import util
from beir.datasets.data_loader import GenericDataLoader

for name in ["fiqa", "scidocs", "climate-fever", "hotpotqa"]:
    url = f"https://public.ukp.informatik.tu-darmstadt.de/thakur/BEIR/datasets/{name}.zip"
    util.download_and_unzip(url, "/kaggle/working/data")

!rm /kaggle/working/data/*.zip
```

```python
for name in ["fiqa", "scidocs", "climate-fever", "hotpotqa"]:
    corpus, queries, qrels = GenericDataLoader(f"/kaggle/working/data/{name}").load(split="test")
    print(name, len(corpus), len(queries))
```
